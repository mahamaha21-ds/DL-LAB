```python

a = np.array([[3, 5],
              [4, 2]])

eigenvalues, eigenvectors = np.linalg.eig(a)

# Create diagonal matrix from eigenvalues
D = np.diag(eigenvalues)

# Inverse of the eigenvector matrix
V = eigenvectors
V_inv = np.linalg.inv(V)

# Reconstruct the original matrix using eigen decomposition
a_reconstructed = V @ D @ V_inv


print("Eigenvalues:")
print(eigenvalues)

print("Eigenvectors:")
print(eigenvectors)

print("Reconstructed matrix from eigen decomposition:")
print(a_reconstructed)
```

**EX 3 / SINGULAR VALUE DECOMPOSTION (SVD)**

```python
a=np.array([[3,2,2],[2,3,-2]])
U,S,V=np.linalg.svd(a)
print("original matrix:")
print(a)
print("left singular vector:")
print(U)
print("singular values :")
print(S)
print("right singular value :")
print(V)
```

**EX 4 / DETERMINANT CALCULATION**

**NORM CALCULATION**

```python
v=np.array([3,4])
L1_norm=np.linalg.norm(v,ord=1)
L2_norm=np.linalg.norm(v)
print("vector v:",v)
print("L1 norm of v:",L1_norm)
print("L2 norm of v:",L2_norm)
```
